# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
import re
import json

# -----------------------------
# TOOL 1: Calculator
# -----------------------------
def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"


# -----------------------------
# TOOL 2: Keyword Extractor
# -----------------------------
def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []


# -----------------------------
# Intent Detection
# -----------------------------
def detect_intent(query: str) -> str:
    query = query.lower()

    # Keyword extraction
    if "keyword" in query or "extract" in query:
        return "keyword"

    # Math query
    if re.fullmatch(r"[0-9+\-*/().\s]+", query):
        return "math"

    if any(op in query for op in ["+", "-", "*", "/", "%"]):
        return "math"

    return "general"


# -----------------------------
# Single Agent
# -----------------------------
def smart_agent(query: str) -> dict:

    intent = detect_intent(query)

    if intent == "math":
        result = calculator(query)

        return {
            "intent": "math",
            "tool_used": "calculator",
            "output": result,
            "status": "success" if result != "Error in calculation" else "error"
        }

    elif intent == "keyword":
        # Remove instruction words before extraction
        clean_text = (
            query.replace("extract keywords", "")
                 .replace("extract", "")
                 .replace("keywords", "")
                 .strip()
        )

        result = extract_keywords(clean_text)

        return {
            "intent": "keyword",
            "tool_used": "extract_keywords",
            "output": result,
            "status": "success"
        }

    else:
        return {
            "intent": "general",
            "tool_used": None,
            "output": f"General Response: {query}",
            "status": "success"
        }


# -----------------------------
# Test Examples
# -----------------------------
queries = [
    "25 + 15 * 2",
    "Extract keywords from Artificial Intelligence is transforming Healthcare and Education",
    "What is Machine Learning?"
]

for q in queries:
    print(json.dumps(smart_agent(q), indent=4))
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
import re

def agent(query: str):
    query_lower = query.lower()

    # Route to Calculator
    if "calculate" in query_lower:
        # Extract the mathematical expression
        expression = query_lower.replace("calculate", "").strip()

        return {
            "type": "calculator",
            "result": calculator(expression)
        }

    # Route to Keyword Extractor
    elif "keywords" in query_lower:
        # Remove the command before extracting keywords
        text = query_lower.replace("keywords", "").strip()

        return {
            "type": "keywords",
            "result": extract_keywords(text)
        }

    # General Response
    else:
        return {
            "type": "general",
            "result": f"General response: {query}"
        }
````
